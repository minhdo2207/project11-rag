Minh hoạ yêu cầu non-functional: cùng business logic, đổi backend chỉ ở **1 dòng**.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

from src.models import Document
from src.repository import InMemoryKnowledgeRepository, FileKnowledgeRepository

def run(repo):
    repo.clear()
    repo.add(Document('d1', 'eval(base64_payload)', embedding=[1,0,0], label='malicious'))
    repo.add(Document('d2', 'print("hi")', embedding=[0,1,0], label='benign'))
    return [d.doc_id for d in repo.search([1,0,0], top_k=1)]

# Đổi giữa hai dòng dưới đây — kết quả giống nhau, không sửa hàm run():
print('memory:', run(InMemoryKnowledgeRepository()))
print('file  :', run(FileKnowledgeRepository(path='../data/kb')))
```
